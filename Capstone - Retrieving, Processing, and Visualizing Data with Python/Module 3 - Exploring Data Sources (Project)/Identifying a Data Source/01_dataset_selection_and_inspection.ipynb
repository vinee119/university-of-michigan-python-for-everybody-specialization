## Loading the Dataset

In this section, we load the Open Food Facts dataset into a Pandas DataFrame. This allows us to inspect the dataset and understand its overall structure before performing any preprocessing or analysis.

```python
# Load the dataset

food_df = pd.read_csv("../data/raw/en.openfoodfacts.org.products.tsv", sep = "\t", low_memory = False)
```

## Previewing the Dataset

After loading the dataset successfully, the next step is to examine a sample of the records. Previewing the dataset helps us understand its structure, identify the available variables, and verify that the data has been loaded correctly.

```python
# Display the first five rows of the dataset

food_df.head()
```

## Examining the Dataset Dimensions

Understanding the size of the dataset is an important first step in exploratory data analysis. In this section, we determine the total number of rows and columns to get an overview of the dataset's scale.

```python
# Display the number of rows and columns

food_df.shape
```

## Reviewing Dataset Information

To better understand the dataset, we examine the column names, data types, and the number of non-null values. This information helps identify missing data, understand the structure of the dataset, and determine the preprocessing steps required before analysis.

```python
# Display dataset information

food_df.info()
```

## Examining the Column Names

The dataset contains a large number of variables describing different aspects of food products, including product details, nutritional information, ingredients, categories, and labels. Listing the column names provides an overview of the available features and helps identify the variables that may be useful for further analysis.

```python
# Display all column names

list(food_df.columns)
```

## Assessing Missing Values

Real-world datasets often contain missing or incomplete information. Before performing any analysis, it is important to identify the extent of missing values across the dataset. This helps determine which variables require cleaning, imputation, or removal during the preprocessing stage.

```python
# Calculate the number of missing values in each column

missing_values = food_df.isnull().sum()
missing_values
```

## Identifying Columns with Missing Values

To better understand the completeness of the dataset, we identify only those columns that contain missing values. This provides a clearer view of the variables that may require preprocessing before further analysis.

```python
# Display only the columns that contain missing values

missing_values[missing_values > 0]
```

## Ranking Missing Values by Column

Sorting the missing values in descending order makes it easier to identify the columns with the highest proportion of missing data. This information will guide future preprocessing decisions, such as removing highly incomplete variables or applying appropriate data cleaning techniques.

```python
# Sort missing values in descending order

missing_values.sort_values(ascending = False)
```

## Checking for Duplicate Records

Duplicate records can introduce bias and affect the accuracy of the analysis. In this step, we check whether the dataset contains duplicate entries so that appropriate preprocessing decisions can be made in the next phase of the project.

```python
# Count the number of duplicate records

duplicate_records = food_df.duplicated().sum()
print(f"Number of duplicate records: {duplicate_records}")
```

## Initial Observations

Based on the initial inspection, the Open Food Facts dataset appears to be a large real-world dataset containing detailed information about food products and their nutritional composition. While the dataset includes a rich collection of variables, it also contains a significant number of missing values across several columns. These observations provide a clear direction for the next stage of the project, which focuses on data cleaning and preprocessing before conducting exploratory data analysis.

```python
# Display a summary of the dataset inspection

print(f"Total records        : {food_df.shape[0]:,}")
print(f"Total features       : {food_df.shape[1]}")
print(f"Numerical columns    : {food_df.select_dtypes(include=['number']).shape[1]}")
print(f"Categorical columns  : {food_df.select_dtypes(include=['object']).shape[1]}")
print(f"Duplicate records    : {duplicate_records}")
print(f"Columns with missing values : {(missing_values > 0).sum()}")
```
