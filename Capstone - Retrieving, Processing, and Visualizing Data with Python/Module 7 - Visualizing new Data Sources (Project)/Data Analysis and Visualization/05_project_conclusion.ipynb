# Project Conclusion

This notebook brings together the work completed throughout the OpenFoodFacts Global Nutrition Analysis project. It summarizes the dataset, the preparation and cleaning process, the main findings from exploratory data analysis, and the insights obtained from the visual analysis.

The purpose of this final notebook is to present the overall outcome of the project in a clear and concise way without introducing any new analysis.

## Project Overview

The OpenFoodFacts Global Nutrition Analysis project focuses on exploring nutritional information from a large real-world food product dataset. The project follows a complete data analysis workflow, starting with dataset inspection and preparation, followed by exploratory analysis and visualization of nutritional patterns.

The main goal of the project was to understand the composition of the dataset, examine the nutritional characteristics of food products, identify patterns across countries and food classifications, and present the findings through clear visualizations.

The project was completed through four main stages: dataset selection and inspection, data cleaning and preprocessing, exploratory data analysis, and data visualization and insights. This final notebook brings the results from these stages together to provide an overall conclusion.

## Dataset Summary

The project uses the OpenFoodFacts products dataset, which contains detailed information about food products collected from different countries and brands. The original dataset contained 356,027 products and 163 columns, covering product information, ingredients, categories, countries, food classifications, and nutritional attributes.

During the initial inspection, the dataset was found to contain substantial missing values across many columns. Despite this, it provided a wide range of useful information for studying nutritional characteristics and product composition.

After the data preparation stage, the dataset used for analysis contained 356,027 products and 64 columns. The cleaned dataset retained important product, categorical, classification, and nutritional variables required for the subsequent exploratory and visualization stages.

## Data Preparation Summary

The data preparation stage focused on reducing the original dataset to a more manageable and useful set of variables for analysis. Columns with more than 90% missing values were removed, while `potassium_100g` was retained because it represents an important nutritional attribute despite its high level of missing data.

This process reduced the dataset from 163 columns to 64 columns without removing any product records. The remaining data was then examined for missing values, data types, and duplicate records to make sure it was suitable for further analysis.

The final cleaned dataset contained 356,027 products and 64 columns. No duplicate records were found, and the cleaned dataset was saved in the processed data directory for use in the exploratory and visualization notebooks.

## Exploratory Data Analysis Summary

The exploratory analysis provided a detailed view of the cleaned dataset and its nutritional characteristics. Numerical variables were examined using descriptive statistics, missing-value analysis, histograms, boxplots, and correlation analysis. These analyses showed that several nutritional variables have highly skewed distributions and some extreme values.

The correlation analysis also revealed meaningful relationships between nutritional variables. Energy showed a strong positive relationship with fat and saturated fat, while carbohydrates showed a strong positive relationship with sugars. The French and UK nutrition scores were also very strongly correlated.

Categorical analysis was used to examine the distribution of countries, brands, food categories, nutrition grades, and PNNS food groups. The United States was the most represented country, Carrefour was the most represented brand, and nutrition grade D was the most common grade. Both PNNS Level 1 and Level 2 classifications were also examined to understand the broad and more detailed food-group composition of the dataset.

Overall, the exploratory analysis provided the foundation for selecting the nutritional patterns and relationships examined more closely in the visualization stage.

## Visualization and Key Insights

The visualization stage helped turn the findings from the exploratory analysis into clearer comparisons and patterns. The comparison across major countries showed differences in average nutrition scores, while the analysis of major food categories highlighted noticeable differences in their nutritional composition.

The food-category analysis showed that fats had particularly high fat and saturated-fat levels, while sugary snack categories generally had higher energy and sugar levels. Chocolate-based products also showed relatively high saturated-fat levels, and prepared meats showed a relatively high protein level.

The comparison across PNNS Level 1 food groups showed distinct nutritional profiles. Fat and sauces had high relative fat and saturated-fat levels, sugary snacks had notably high relative sugar levels, and the Fish Meat Eggs group showed a relatively high protein level. Fruits and vegetables generally showed lower relative levels across several of the selected nutritional characteristics.

The relationship analysis also showed clear positive relationships between energy and fat, fat and saturated fat, and carbohydrates and sugars. At the same time, the scatter plots showed variation and some extreme observations within the nutritional data.

Overall, the visualizations made the nutritional differences and relationships in the dataset easier to understand and provided a clearer view of the main patterns identified during the project.

## Overall Findings

The analysis shows that the OpenFoodFacts dataset contains a wide variety of food products with substantial differences in their nutritional characteristics. The dataset also represents products from many countries, brands, categories, and food-group classifications.

The cleaning and exploratory stages made it possible to work with a more focused set of important variables while retaining all product records. The subsequent analysis showed that nutritional composition varies considerably across different types of food products and food classifications.

Some of the clearest patterns were found among energy, fat, saturated fat, sugars, and protein. Products from categories such as fats, sauces, sugary snacks, and salty snacks generally showed higher levels of energy, fat, and other related nutritional characteristics, while groups such as fruits and vegetables tended to show lower relative levels. The analysis also highlighted products and groups with comparatively higher protein levels.

The project also showed that several nutritional variables are closely related. In particular, energy and fat, fat and saturated fat, and carbohydrates and sugars showed clear positive relationships. These relationships help explain how different nutritional characteristics tend to occur together across food products.

Overall, the project provided a broad view of the nutritional composition of the dataset and showed how data cleaning, exploratory analysis, and visualization can be combined to identify meaningful patterns in a large real-world dataset.
